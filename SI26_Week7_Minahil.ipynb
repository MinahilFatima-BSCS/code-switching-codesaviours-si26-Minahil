```python
import pandas as pd

df = pd.read_csv("/content/dataset.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head(10))
```

```python
print("Total rows:", len(df))

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate sentences:")
print(df["sentence"].duplicated().sum())
```

```python
print("Label distribution:")
print(df["label"].value_counts())

print("\nUnique labels:")
print(df["label"].unique())
```

```python
display(
    df.groupby("label")["word"]
      .count()
      .reset_index(name="word_count")
)
```

```python
print("Unique labels:")
print(df["label"].unique())

print("\nSample ENG words:")
display(df[df["label"] == "ENG"].head(15))

print("\nSample URD words:")
display(df[df["label"] == "URD"].head(15))
```

```python
# Show all unique words currently labeled as URD
urd_words = sorted(
    df[df["label"] == "URD"]["word"]
    .astype(str)
    .str.lower()
    .unique()
)

print("Total unique URD words:", len(urd_words))
print(urd_words[:200])
```

```python
# Show all unique words currently labeled as ENG
eng_words = sorted(
    df[df["label"] == "ENG"]["word"]
    .astype(str)
    .str.lower()
    .unique()
)

print("Total unique ENG words:", len(eng_words))
print(eng_words[:200])
```

```python
word_label_counts = (
    df.groupby("word")["label"]
      .nunique()
      .sort_values(ascending=False)
)

ambiguous_words = word_label_counts[word_label_counts > 1]

print("Words having multiple labels:", len(ambiguous_words))

display(
    df[df["word"].isin(ambiguous_words.index)]
    .sort_values("word")
    [["sentence", "word", "label"]]
    .head(100)
)
```

```python
# Group word-level rows into sentence-level records

sentence_data = (
    df.groupby("sentence", sort=False)
      .agg(
          words=("word", list),
          labels=("label", list)
      )
      .reset_index()
)

print("Total unique sentences:", len(sentence_data))
print("Columns:", sentence_data.columns.tolist())

display(sentence_data.head(10))
```

```python
# Verify that every sentence has matching number of words and labels

sentence_data["word_count"] = sentence_data["words"].apply(len)
sentence_data["label_count"] = sentence_data["labels"].apply(len)

print(
    "Mismatched word/label counts:",
    (sentence_data["word_count"] != sentence_data["label_count"]).sum()
)
```

```python
from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(
    sentence_data,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

train_data = train_data.reset_index(drop=True)
test_data = test_data.reset_index(drop=True)

print("Training sentences:", len(train_data))
print("Testing sentences:", len(test_data))

print("\nTraining percentage:",
      len(train_data) / len(sentence_data) * 100)

print("Testing percentage:",
      len(test_data) / len(sentence_data) * 100)
```

```python
print("TRAINING DATA")
display(train_data.head(5))

print("\nTESTING DATA")
display(test_data.head(5))
```

```python
from collections import Counter

train_label_counts = Counter(
    label
    for labels in train_data["labels"]
    for label in labels
)

test_label_counts = Counter(
    label
    for labels in test_data["labels"]
    for label in labels
)

print("Training labels:")
print(train_label_counts)

print("\nTesting labels:")
print(test_label_counts)
```

```python
label2id = {
    "URD": 0,
    "ENG": 1
}

id2label = {
    0: "URD",
    1: "ENG"
}

print("label2id:", label2id)
print("id2label:", id2label)
```

```python
from transformers import AutoTokenizer

model_name = "xlm-roberta-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)

print("Tokenizer loaded successfully!")
print("Model:", model_name)
print("Vocabulary size:", tokenizer.vocab_size)
```

```python
def tokenize_and_align_labels(examples):

    tokenized = tokenizer(
        examples["words"],
        truncation=True,
        is_split_into_words=True
    )

    all_labels = []

    for i, label_list in enumerate(examples["labels"]):

        word_ids = tokenized.word_ids(batch_index=i)

        label_ids = []
        previous_word_id = None

        for word_id in word_ids:

            # Special token
            if word_id is None:
                label_ids.append(-100)

            # First token of a word
            elif word_id != previous_word_id:
                label_ids.append(
                    label2id[label_list[word_id]]
                )

            # Additional sub-token
            else:
                label_ids.append(-100)

            previous_word_id = word_id

        all_labels.append(label_ids)

    tokenized["labels"] = all_labels

    return tokenized
```

```python
from datasets import Dataset

train_ds = Dataset.from_pandas(
    train_data[["words", "labels"]],
    preserve_index=False
)

test_ds = Dataset.from_pandas(
    test_data[["words", "labels"]],
    preserve_index=False
)

print("Training Dataset:")
print(train_ds)

print("\nTesting Dataset:")
print(test_ds)
```

```python
train_tokenized = train_ds.map(
    tokenize_and_align_labels,
    batched=True
)

test_tokenized = test_ds.map(
    tokenize_and_align_labels,
    batched=True
)

print("Training tokenization completed!")
print("Testing tokenization completed!")

print("\nTraining samples:", len(train_tokenized))
print("Testing samples:", len(test_tokenized))
```

```python
print("Original words:")
print(train_ds[0]["words"])

print("\nOriginal labels:")
print(train_ds[0]["labels"])

print("\nToken IDs:")
print(train_tokenized[0]["input_ids"])

print("\nAligned labels:")
print(train_tokenized[0]["labels"])
```

```python
from transformers import AutoModelForTokenClassification

model = AutoModelForTokenClassification.from_pretrained(
    model_name,
    num_labels=2,
    id2label=id2label,
    label2id=label2id
)

print("Model loaded successfully!")
print("Model:", model_name)
print("Number of labels:", model.config.num_labels)
print("Labels:", model.config.id2label)
```

```python
import torch

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

model.to(device)
```

```python
import numpy as np
from sklearn.metrics import f1_score, accuracy_score

print("Evaluation metrics ready!")
```

```python
def compute_metrics(p):

    predictions, labels = p

    predictions = np.argmax(predictions, axis=2)

    true_predictions = []
    true_labels = []

    for prediction, label in zip(predictions, labels):

        for pred, lab in zip(prediction, label):

            if lab != -100:
                true_predictions.append(pred)
                true_labels.append(lab)

    f1_urd = f1_score(
        true_labels,
        true_predictions,
        labels=[0],
        average="macro",
        zero_division=0
    )

    f1_eng = f1_score(
        true_labels,
        true_predictions,
        labels=[1],
        average="macro",
        zero_division=0
    )

    overall_f1 = f1_score(
        true_labels,
        true_predictions,
        average="macro",
        zero_division=0
    )

    accuracy = accuracy_score(
        true_labels,
        true_predictions
    )

    return {
        "f1_URD": f1_urd,
        "f1_ENG": f1_eng,
        "overall_f1": overall_f1,
        "overall_accuracy": accuracy
    }

print("F1 evaluation function ready!")
```

```python
from transformers import DataCollatorForTokenClassification

data_collator = DataCollatorForTokenClassification(
    tokenizer=tokenizer
)

print("Data collator created successfully!")
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_steps=10,
    load_best_model_at_end=True,
    metric_for_best_model="overall_f1",
    greater_is_better=True,
    report_to="none",
    fp16=True
)

print("Training arguments configured successfully!")
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokenized,
    eval_dataset=test_tokenized,
    processing_class=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)

print("Trainer created successfully!")
```

```python
print("Starting XLM-RoBERTa training...")

train_result = trainer.train()

print("\nTraining completed successfully!")
```

```python
evaluation_results = trainer.evaluate()

print("\n===== FINAL EVALUATION =====")

for key, value in evaluation_results.items():
    print(f"{key}: {value}")
```

```python
from sklearn.metrics import classification_report

predictions = trainer.predict(test_tokenized)

pred_ids = np.argmax(
    predictions.predictions,
    axis=2
)

true_ids = predictions.label_ids

y_true = []
y_pred = []

for pred_row, true_row in zip(pred_ids, true_ids):

    for pred, true in zip(pred_row, true_row):

        if true != -100:
            y_true.append(true)
            y_pred.append(pred)

print(
    classification_report(
        y_true,
        y_pred,
        labels=[0, 1],
        target_names=["URD", "ENG"],
        digits=4,
        zero_division=0
    )
)
```

```python
save_path = "./urdu-eng-xlm-roberta"

trainer.save_model(save_path)
tokenizer.save_pretrained(save_path)

print("Model saved successfully!")
print("Saved at:", save_path)
```
